```python
# union_conjuntos.py
def union_conjuntos():
    """Programa para encontrar la unión de dos conjuntos"""

    print("       UNIÓN DE CONJUNTOS (A ∪ B)")

    print("Elementos que se encuentran en A o en B o en ambos")
    print()

    # Ingresar conjunto A
    print("Ingrese los elementos del conjunto A (separados por espacios):")
    elementos_a = input().split()
    conjunto_a = set(elementos_a)

    # Ingresar conjunto B
    print("Ingrese los elementos del conjunto B (separados por espacios):")
    elementos_b = input().split()
    conjunto_b = set(elementos_b)

    # Calcular unión
    union = conjunto_a | conjunto_b

    # Mostrar resultados

    print("RESULTADOS:")
    print(f"Conjunto A: {conjunto_a}")
    print(f"Conjunto B: {conjunto_b}")
    print(f"Unión (A ∪ B): {union}")
    print(f"Número de elementos en la unión: {len(union)}")
    print("=" * 50)

if __name__ == "__main__":
    union_conjuntos()
```

```python
# Programa para practicar operaciones con conjuntos

def mostrar_menu():
    """Muestra el menú principal"""
    print("\n" + "="*50)
    print("       PRÁCTICA DE OPERACIONES CON CONJUNTOS")
    print("="*50)
    print("1. Unión (A ∪ B) - Elementos en A o B o ambos")
    print("2. Intersección (A ∩ B) - Elementos en A y B")
    print("3. Diferencia simétrica (A Δ B) - Elementos en A o B pero no en ambos")
    print("4. Verificar subconjunto (A ⊆ B)")
    print("5. Contar elementos de un conjunto")
    print("6. Ingresar nuevos conjuntos")
    print("7. Mostrar conjuntos actuales")
    print("8. Salir")
    print("="*50)

def ingresar_conjunto(nombre):
    """Solicita al usuario ingresar un conjunto"""
    print(f"\nIngrese los elementos del conjunto {nombre} (separados por espacios):")
    elementos = input().split()
    # Convertimos a conjunto y manejamos números y strings
    conjunto = set()
    for elemento in elementos:
        # Intentamos convertir a número si es posible
        try:
            if '.' in elemento:
                conjunto.add(float(elemento))
            else:
                conjunto.add(int(elemento))
        except ValueError:
            conjunto.add(elemento)
    return conjunto

def operacion_union(conjunto_a, conjunto_b):
    """Realiza y muestra la unión de dos conjuntos"""
    resultado = conjunto_a | conjunto_b
    print(f"\nUnión (A ∪ B): {resultado}")
    return resultado

def operacion_interseccion(conjunto_a, conjunto_b):
    """Realiza y muestra la intersección de dos conjuntos"""
    resultado = conjunto_a & conjunto_b
    print(f"\nIntersección (A ∩ B): {resultado}")
    return resultado

def operacion_diferencia_simetrica(conjunto_a, conjunto_b):
    """Realiza y muestra la diferencia simétrica de dos conjuntos"""
    resultado = conjunto_a ^ conjunto_b
    print(f"\nDiferencia simétrica (A Δ B): {resultado}")
    return resultado

def verificar_subconjunto(conjunto_a, conjunto_b):
    """Verifica si A es subconjunto de B"""
    es_subconjunto = conjunto_a.issubset(conjunto_b)
    print(f"\n¿A ⊆ B? {es_subconjunto}")

    if es_subconjunto:
        print(f"Todos los elementos de A ({conjunto_a}) están en B ({conjunto_b})")
    else:
        elementos_faltantes = conjunto_a - conjunto_b
        print(f"Elementos de A que no están en B: {elementos_faltantes}")

    return es_subconjunto

def contar_elementos(conjunto):
    """Cuenta y muestra el número de elementos de un conjunto"""
    cantidad = len(conjunto)
    print(f"\nEl conjunto tiene {cantidad} elementos: {conjunto}")
    return cantidad

def mostrar_conjuntos(conjunto_a, conjunto_b):
    """Muestra los conjuntos actuales"""
    print(f"\nConjunto A: {conjunto_a}")
    print(f"Conjunto B: {conjunto_b}")
    print(f"Tipo de elementos en A: {type(next(iter(conjunto_a))) if conjunto_a else 'Vacío'}")
    print(f"Tipo de elementos en B: {type(next(iter(conjunto_b))) if conjunto_b else 'Vacío'}")

def main():
    """Función principal del programa"""
    print("BIENVENIDO AL PRACTICADOR DE CONJUNTOS")
    print("Primero, ingrese los conjuntos A y B:")

    # Inicializar conjuntos
    conjunto_a = ingresar_conjunto("A")
    conjunto_b = ingresar_conjunto("B")

    while True:
        mostrar_conjuntos(conjunto_a, conjunto_b)
        mostrar_menu()

        opcion = input("\nSeleccione una opción (1-8): ").strip()

        if opcion == '1':
            operacion_union(conjunto_a, conjunto_b)

        elif opcion == '2':
            operacion_interseccion(conjunto_a, conjunto_b)

        elif opcion == '3':
            operacion_diferencia_simetrica(conjunto_a, conjunto_b)

        elif opcion == '4':
            verificar_subconjunto(conjunto_a, conjunto_b)

        elif opcion == '5':
            print("\n¿De qué conjunto desea contar los elementos?")
            print("a) Conjunto A")
            print("b) Conjunto B")
            print("c) Ambos conjuntos")

            sub_opcion = input("Seleccione (a/b/c): ").lower().strip()

            if sub_opcion == 'a':
                contar_elementos(conjunto_a)
            elif sub_opcion == 'b':
                contar_elementos(conjunto_b)
            elif sub_opcion == 'c':
                print(f"\nConjunto A: {len(conjunto_a)} elementos")
                print(f"Conjunto B: {len(conjunto_b)} elementos")
                print(f"Total único: {len(conjunto_a | conjunto_b)} elementos")
            else:
                print("Opción no válida")

        elif opcion == '6':
            print("\n¿Qué conjunto desea modificar?")
            print("a) Conjunto A")
            print("b) Conjunto B")
            print("c) Ambos conjuntos")

            sub_opcion = input("Seleccione (a/b/c): ").lower().strip()

            if sub_opcion == 'a':
                conjunto_a = ingresar_conjunto("A")
            elif sub_opcion == 'b':
                conjunto_b = ingresar_conjunto("B")
            elif sub_opcion == 'c':
                conjunto_a = ingresar_conjunto("A")
                conjunto_b = ingresar_conjunto("B")
            else:
                print("Opción no válida")

        elif opcion == '7':
            mostrar_conjuntos(conjunto_a, conjunto_b)

        elif opcion == '8':
            print("\n¡Gracias por practicar! ¡Hasta pronto!")
            break

        else:
            print("Opción no válida. Por favor, seleccione 1-8.")

        # Pausa antes de continuar
        input("\nPresione Enter para continuar...")

# Ejemplos de prueba por si el usuario necesita ayuda
def mostrar_ejemplos():
    """Muestra ejemplos de uso"""
    print("\n" + "="*60)
    print("EJEMPLOS DE USO:")
    print("="*60)

    # Ejemplo 1: Conjuntos numéricos
    print("\n1. Conjuntos numéricos:")
    A = {1, 2, 3, 4, 5}
    B = {4, 5, 6, 7, 8}
    print(f"A = {A}")
    print(f"B = {B}")
    print(f"Unión (A | B): {A | B}")
    print(f"Intersección (A & B): {A & B}")
    print(f"Diferencia simétrica (A ^ B): {A ^ B}")

    # Ejemplo 2: Conjuntos de strings
    print("\n2. Conjuntos de strings:")
    C = {"manzana", "banana", "naranja"}
    D = {"banana", "uva", "pera"}
    print(f"C = {C}")
    print(f"D = {D}")
    print(f"Unión: {C | D}")

    # Ejemplo 3: Verificación de subconjunto
    print("\n3. Verificación de subconjunto:")
    E = {1, 2, 3}
    F = {1, 2, 3, 4, 5}
    print(f"E = {E}")
    print(f"F = {F}")
    print(f"¿E es subconjunto de F? {E.issubset(F)}")

if __name__ == "__main__":
    # Mostrar ejemplos primero
    mostrar_ejemplos()

    # Preguntar si quieren ver los ejemplos o ir directamente al programa
    respuesta = input("\n¿Desea ver más ejemplos antes de comenzar? (s/n): ").lower().strip()

    if respuesta == 's':
        print("\nMás ejemplos:")
        print("Puede ingresar conjuntos mixtos como: '1 2.5 manzana True'")
        print("Los números se convertirán automáticamente a int o float")
        print("Los strings se mantendrán como texto")

    # Iniciar el programa principal
    main()
```
